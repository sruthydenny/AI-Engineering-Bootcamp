Write and test a function named 'cos_similarity_calculation' that determines the semantic similarity between the sentences in your list and any text sentence using their vector representations and the cosine distance as a similarity measure.  

```python
def cos_similarity_calculation(reference_sentence, candidate_sentences, model):
    reference_embedding = model.encode(reference_sentence, convert_to_tensor=True)
    candidate_embeddings = model.encode(candidate_sentences, convert_to_tensor=True)
    return util.cos_sim(reference_embedding, candidate_embeddings)

query_sentence = "People use digital tools to coordinate work and share information across teams."
similarity_scores = cos_similarity_calculation(query_sentence, sentences_list, model)
similarity_scores
```

Create a function that determines the cosine similarity between a vector and a batch of vectors using the cosine distance formula and the numpy library. Add code to demonstrate how to use this function.

```python
import numpy as np

def cosine_similarity_matrix(vector, vectors):
    vector = np.asarray(vector, dtype=np.float32)
    vectors = np.asarray(vectors, dtype=np.float32)
    vector_norm = vector / np.linalg.norm(vector)
    vectors_norm = vectors / np.linalg.norm(vectors, axis=1, keepdims=True)
    return vectors_norm @ vector_norm

query_vector = model.encode("Machine learning systems can identify patterns in large amounts of data.", convert_to_tensor=False)
batch_vectors = model.encode(sentences_list, convert_to_tensor=False)
similarities = cosine_similarity_matrix(query_vector, batch_vectors)
similarities
```
