Task 8

 Identify potential outliers in `alcohol` using the IQR rule. Flag values below `Q1 - 1.5 × IQR` or above `Q3 + 1.5 × IQR`.

```python
q1 = wine_data['alcohol'].quantile(0.25)
q3 = wine_data['alcohol'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

alcohol_outliers = wine_data.loc[
    (wine_data['alcohol'] < lower_fence) | (wine_data['alcohol'] > upper_fence),
    ['alcohol']
].sort_values('alcohol')

print(f"Lower fence: {lower_fence:,.2f}")
print(f"Upper fence: {upper_fence:,.2f}")
alcohol_outliers
```

Task 13

 Create X, y for independent variables (all except `quality`) and target variable `quality`, perform k-fold k=3 split on them and calculate and print out the length of training set and testing set for each fold


```python
X = wine_data.drop('quality', axis=1)  # independent variables
y = wine_data['quality']              # target variable
```

```python
k = 3

kf = KFold(n_splits=k, shuffle=True, random_state=42)
```

```python
fold = 1

for train_index, test_index in kf.split(X):

    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    print(f"Fold {fold}:")
    print("Training set length:", len(X_train))
    print("Testing set length:", len(X_test))
    print()

    fold += 1
```

Task 15

 Create a sample dataset with dates for a whole week. Extract day of the week with `dayofweek` function (0=Monday, 6=Sunday). Additionally flag weekends by creating another new `is_weekend` feature.

```python

date_data = {
    'date': [
        '2023-11-06', '2023-11-07', '2023-11-08',
        '2023-11-09', '2023-11-10', '2023-11-11',
        '2023-11-12'
    ]
}
df = pd.DataFrame(date_data)
df['date'] = pd.to_datetime(df['date'])
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day'] = df['date'].dt.day
df['day_of_week'] = df['date'].dt.dayofweek
df['is_weekend'] = df['day_of_week'].apply(lambda x: 1 if x >= 5 else 0)
df['quarter'] = df['date'].dt.quarter
df['day_of_year'] = df['date'].dt.dayofyear
df
```
